## **2.5 List Methods**

```python
#2.5 Create a list of 5 numbers, add a new number, remove one, and then sort the list

number = ["10", "6", "12", "2", "56"]
print(number)

#adding a new number
number.append('20')
print(number)

#adding a new number at position
number.insert(1, '25')
print(number)

#removing elements
number.remove('12')
print(number)
popped=number.pop()
print(number)
print("Popped item is ",popped)

#sorting element

number.sort() #sort as alphabetical order
print(number)
number.reverse() #sort as reverse order
print(number)
```

##**2.8 Sets**

```python
# 2.8 create two sets of color and find their union and intersection

color1 = {"Green","Red","Yellow","Orange",}
color2 = {"Purple","Red","Blue","Yellow","Black",}
print(color1, color2)

#union of sets by using '|'
All_color = color1 | color2
print(All_color)

#intersection of sets by using '&'
common_color = color1 & color2
print(common_color)
```

##**2.10 Nested If**

```python
# 2.10 Write a program that checks eligibility for a discount (student AND under 25, OR senior over 60)

is_student=True
age = input()

if age<'25' and is_student:
  print("Eligible for Discount")
elif age>'60':
  print("Eligible for Discount")
else:
  print("Not eligible for Discount")
```
